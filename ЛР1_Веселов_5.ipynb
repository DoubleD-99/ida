# Загрузка набора данных

Набор данных: `visits.csv` — данные пользовательских сессий интернет-магазина.

Атрибуты:
- `user_id`: уникальный идентификатор пользователя
- `region`: страна пользователя
- `device`: устройство пользователя

### 1.Чтение файла (набора данных)

### 3. Проверка пропусков

```python
# 1 Проверка технических пропусков (NaN)
print("Проверка технических пропусков (NaN):")
na_counts = df.isna().sum()
if na_counts.sum() == 0:
    print("Технические пропуски не обнаружены.")
else:
    print("Технические пропуски:")
    print(na_counts[na_counts > 0])

# 2 Проверка скрытых/семантических пропусков (пустые строки)
print("\nПроверка скрытых пропусков (пустые строки):")
hidden_missing_found = False
for col in df.columns:
    empty_count = (df[col].astype(str).str.strip() == '').sum()
    if empty_count > 0:
        print(f"️ Столбец '{col}': найдено {empty_count} пустых значений!")
        hidden_missing_found = True

if not hidden_missing_found:
    print("Скрытые пропуски не обнаружены.")

# 3 Анализ уникальных значений в категориальных столбцах
print("\nУникальные значения категориальных столбцов:")
print(f"region: {sorted(df['region'].dropna().unique())}")
print(f"device: {sorted(df['device'].dropna().unique())}")
print(f"channel: {sorted(df['channel'].dropna().unique())}")

# 4 Проверка на отрицательные значения
print("\nПроверка на отрицательные значения:")
non_negative_cols = ['time_session', 'click_count', 'buy_count', 'price', 'age']
negative_found = False

for col in non_negative_cols:
    neg_count = (df[col] < 0).sum()
    if neg_count > 0:
        print(f"Столбец '{col}': найдено {neg_count} отрицательных значений!")
        print(df[df[col] < 0][[col]].head())
        negative_found = True

if not negative_found:
    print("Отрицательные значения в числовых столбцах не обнаружены.")
```

```python
# Замена пропусков в region и device на 'Unknown'.
df['device'] = df['device'].fillna('Unknown')
df['region'] = df['region'].fillna('Unknown')

# Проверка после обработки
print("\nКоличество пропусков после обработки:")
print(df.isna().sum())
```


---

**Вывод:** Пропуски в категориальных столбцах `region` и `device` были заполнены значением 'Unknown'. Удаление строк привело бы к потере данных о сессиях, которые могут содержать важную информацию о покупках или времени. Замена на модальное значение (самое частое) исказила бы статистику по регионам/устройствам, поэтому выбран нейтральный вариант 'Unknown'. Отрицательные значения в числовых столбцах не обнаружены.


 ---

### 4. Проверка дубликатов

#### Проверка явных дубликатов

```python
# Поиск полных дубликатов строк
duplicates_count = df.duplicated().sum()
print(f"Количество явных дубликатов: {duplicates_count}")

if duplicates_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print("Дубликаты удалены.")
else:
    print("Явных дубликатов нет.")
```

#### Проверка неявных дубликатов

```python
# Проверка уникальных значений в категориальных столбцах на предмет опечаток
print("Уникальные значения device:")
print(df['device'].value_counts())

print("\nУникальные значения region:")
print(df['region'].value_counts())

print("\nУникальные значения channel:")
print(df['channel'].value_counts())
```

```python
# Исправление неявных дубликатов
# Приводим первую букву каждой строки к верхнему регистру, а все остальные — к нижнему
df['device'] = df['device'].str.capitalize()
# Корректировка конкретных случаев
df['device'] = df['device'].replace({'Iphone': 'iPhone', 'Pc': 'PC'})

# Проверка region
df['region'] = df['region'].replace({'USA': 'United States'})

print(f"Уникальные значения device после обработки: {df['device'].unique()}")
print(f"Уникальные значения region после обработки: {df['region'].unique()}")
```

---

**Вывод:** Были выявлены неявные дубликаты в названиях устройств ('IPHONE' vs 'iPhone', 'MAC' vs 'Mac'), регионов ('USA' vs 'United States'). Произведена нормализация регистра и замена синонимов для обеспечения целостности данных при группировке.


 ---

### 5. Провека типов данных

```python
print("Типы данных до преобразования:")
print(df.dtypes)
```

```python
# Преобразование типов
# user_id лучше оставить как object, так это метка
df['user_id'] = df['user_id'].astype(str)

# Даты
df['session_start'] = pd.to_datetime(df['session_start'], format='%d.%m.%Y %H:%M', errors='coerce')
df['session_end'] = pd.to_datetime(df['session_end'], format='%d.%m.%Y %H:%M', errors='coerce')

# click_count и buy_count целые числа
df['click_count'] = df['click_count'].astype('int64')
df['buy_count'] = df['buy_count'].astype('int64')

print("Типы данных после преобразования:")
print(df.dtypes)
```

---

**Вывод:** `user_id` переведен в объект, поскольку является меткой, а не числом.
Столбцы с датами переведены в формат `datetime64` для возможности вычисления длительности и фильтрации по времени. Числовые столбцы приведены к числовым типам (float/int) для проведения математических операций.

### 6. Группировка данных

#### Задание 1

Группировка - `device` и количество рекламных источников каждого типа (`channel`).

```python
task1 = df.groupby('device')['channel'].value_counts()
print(task1)
```

**Вывод:** Из таблицы видно распределение каналов привлечения трафика для каждого типа устройства. Например, пользователи iPhone чаще всего приходят из канала 'organic'. Это позволяет понять, какие каналы эффективны для разных платформ.

#### Задание 2

Группировка - region и количество устройств (device ).
Создать датафрейм. Переименовать столбец с количеством в “сount”.
Отсортировать по убыванию столбца “count”

```python
task2 = df.groupby(['region', 'device']).size().reset_index(name='count')
task2 = task2.sort_values(by='count', ascending=False).reset_index(drop=True)
print(task2.head(10))
```

**Вывод:** Наибольшее количество сессий зафиксировано в регионе United States, преимущественно с устройств iPhone. Россия занимает второе место, где также лидируют мобильные устройства. Сортировка по убыванию позволила выявить самые популярные комбинации регион-устройство.


#### Задание 3

Сводная таблица (pivot_table) - уникальное количество
пользователей для каждого устройства (device). Отсортировать по
убыванию количества.

```python
task3 = df.pivot_table(index='device', values='user_id', aggfunc='nunique')
task3 = task3.rename(columns={'user_id': 'unique_users'})
task3 = task3.sort_values(by='unique_users', ascending=False)
print(task3)
```

**Вывод:** iPhone является самым популярным устройством среди уникальных пользователей, за ним следуют Mac и Android. PC показывает наименьшее количество уникальных пользователей. Это указывает на мобильную ориентированность аудитории магазина.


#### Задание 4

Сводная таблица (pivot_table) - уникальное количество
пользователей для каждого устройства (device) - столбцы и регионастроки. Отсортировать по возрастанию столбца region

```python
task4 = df.pivot_table(index='region', columns='device', values='user_id', aggfunc='nunique')
task4 = task4.fillna(0).astype('int64')
task4 = task4.sort_index(axis=0, ascending=True) # cортировка по регионам (строкам)
print(task4)
```

**Вывод:** Сводная таблица показывает распределение уникальной аудитории по странам и типам устройств. В США преобладают iPhone и Mac, в то время как в России доля iPhone еще выше относительно других устройств. Сортировка по алфавиту облегчает поиск конкретного региона.
